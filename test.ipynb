```python
def group_by(column : str ="Department",
                   target : str = "MonthlyIncome",
                   aggregator : str = "mean") :
        agg_result = df.groupby(column)[target].agg(aggregator)
        return agg_result.to_dict()

print(group_by())
```

```python
def filter_by(column : str ="Department",
                    cat : list(str)= ["Sales"]):
         filter_result = df[df[column].isin(cat)]
         return filter_result.to_dict(orient="records")
print(filter_by())
```
